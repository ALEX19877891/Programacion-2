1. Clases abtractas vehiculos

```python
from abc import ABC, abstractmethod

class Vehiculo(ABC):
    def __init__(self, id, modelo, velocidad_max, capacidad):
        self.__id = id
        self.__modelo = modelo
        self.__velocidad_max = velocidad_max
        self.__capacidad = capacidad

    @property
    def id(self):
        return self.__id

    @property
    def modelo(self):
        return self.__modelo

    @property
    def velocidad_max(self):
        return self.__velocidad_max

    @property
    def capacidad(self):
        return self.__capacidad

    @abstractmethod
    def acelerar(self):
        pass

    @abstractmethod
    def frenar(self):
        pass

    def informar_estado(self):
        print(f"Vehículo: {self.modelo}, Velocidad máxima: {self.velocidad_max} km/h")
```

2. Subclases de vehiculo, vehiculo,motociclista

```python
class Automovil(Vehiculo):
    def __init__(self, id, modelo, velocidad_max, capacidad):
        super().__init__(id, modelo, velocidad_max, capacidad)

    def acelerar(self):
        print(f"El automóvil {self.modelo} acelera a {self.velocidad_max} km/h.")

    def frenar(self):
        print(f"El automóvil {self.modelo} frena.")

class Camion(Vehiculo):
    def __init__(self, id, modelo, velocidad_max, capacidad):
        super().__init__(id, modelo, velocidad_max, capacidad)

    def acelerar(self):
        print(f"El camión {self.modelo} acelera a {self.velocidad_max} km/h.")

    def frenar(self):
        print(f"El camión {self.modelo} frena.")

    def engranar_remolque(self):
        print(f"El camión {self.modelo} engranó el remolque.")

class Motocicleta(Vehiculo):
    def __init__(self, id, modelo, velocidad_max, capacidad):
        super().__init__(id, modelo, velocidad_max, capacidad)

    def acelerar(self):
        print(f"La motocicleta {self.modelo} acelera a {self.velocidad_max} km/h.")

    def frenar(self):
        print(f"La motocicleta {self.modelo} frena.")

    def maniobra_evasiva(self):
        print(f"La motocicleta {self.modelo} realiza maniobra evasiva.")
```

3. Implementacion el Strategy Pattern

```python
class EstrategiaConduccion(ABC):
    @abstractmethod
    def conducir(self, vehiculo):
        pass

class ConduccionEconomica(EstrategiaConduccion):
    def conducir(self, vehiculo):
        print(f"Conducción económica para el {vehiculo.modelo}.")

class ConduccionDeportiva(EstrategiaConduccion):
    def conducir(self, vehiculo):
        print(f"Conducción deportiva para el {vehiculo.modelo}.")
```

4.Implementar el Decorator Pattern

```python
class VehiculoDecorator(Vehiculo):
    def __init__(self, vehiculo):
        self._vehiculo = vehiculo

    def acelerar(self):
        return self._vehiculo.acelerar()

    def frenar(self):
        return self._vehiculo.frenar()

class PilotoAutomaticoDecorator(VehiculoDecorator):
    def acelerar(self):
        print("Piloto automático activado.")
        return self._vehiculo.acelerar()

class AsistenteEstacionamientoDecorator(VehiculoDecorator):
    def frenar(self):
        print("Asistente de estacionamiento activado.")
        return self._vehiculo.frenar()
```

5: Implementar el Singleton Pattern

```python
class ControlDeFlota:
    _instancia = None

    def __new__(cls):
        if cls._instancia is None:
            cls._instancia = super(ControlDeFlota, cls).__new__(cls)
            cls._instancia.flota = []
        return cls._instancia

    def agregar_vehiculo(self, vehiculo):
        self.flota.append(vehiculo)

    def eliminar_vehiculo(self, vehiculo):
        self.flota.remove(vehiculo)

    def mostrar_vehiculos(self):
        for vehiculo in self.flota:
            print(f"Vehículo: {vehiculo.modelo}")
```

1. Clase Abstracta Vehiculo

La clase Vehiculo es una clase abstracta que define las características y comportamientos comunes a todos los vehículos, como id, modelo, velocidad_maxima y capacidad_carga. También define métodos abstractos acelerar y frenar que deben ser implementados por las subclases.

2. Subclases de Vehiculo
Las subclases Automovil, Camion y Motocicleta heredan de la clase Vehiculo y implementan los métodos acelerar y frenar de manera específica para cada tipo de vehículo.

La clase EstrategiaConduccion es una clase abstracta que define el comportamiento de conducción de un vehículo. Las subclases ConduccionEconomica y ConduccionDeportiva implementan este comportamiento de manera específica.

Los decoradores PilotoAutomaticoDecorator y AsistenteEstacionamientoDecorator agregan funcionalidades adicionales a los vehículos sin modificar su código original.
